```python
import pandas as pd
import numpy as np

np.random.seed(42)

n = 1000

categories = [
    "Food",
    "Transport",
    "Shopping",
    "Entertainment",
    "Education",
    "Healthcare",
    "Bills"
]

category_names = {
    "Food": ["Canteen", "Swiggy", "Zomato", "Restaurant"],
    "Transport": ["Bus", "Uber", "Ola", "Metro"],
    "Shopping": ["Amazon", "Myntra", "Local Store"],
    "Entertainment": ["Netflix", "PVR", "Gaming"],
    "Education": ["Udemy", "Books", "College"],
    "Healthcare": ["Pharmacy", "Clinic"],
    "Bills": ["Electricity", "Internet", "Mobile"]
}

category_amounts = {
    "Food": (80, 500),
    "Transport": (30, 300),
    "Shopping": (200, 2000),
    "Entertainment": (100, 1000),
    "Education": (100, 1500),
    "Healthcare": (100, 1200),
    "Bills": (200, 1500)
}

dates = pd.date_range(
    start="2026-07-01",
    end="2026-09-30",
    periods=n
)

rows = []

for date in dates:
    category = np.random.choice(categories)

    name = np.random.choice(category_names[category])

    low, high = category_amounts[category]
    amount = np.random.randint(low, high + 1)

    account = np.random.choice(
        ["Savings", "Cash", "Credit"],
        p=[0.6, 0.25, 0.15]
    )

    rows.append([
        date,
        name,
        amount,
        category,
        account,
        "Expense",
        ""
    ])

df = pd.DataFrame(
    rows,
    columns=[
        "date",
        "name",
        "amount",
        "category",
        "account",
        "type",
        "note"
    ]
)

df.head()
```

```python
def analyze_expense(amount, category):
    stats = category_stats.loc[category]

    category_zscore = (
        amount - stats["category_mean"]
    ) / stats["category_std"]

    features = pd.DataFrame(
        [[amount, category_zscore]],
        columns=["amount", "category_zscore"]
    )

    prediction = model_v2.predict(features)[0]
    score = model_v2.decision_function(features)[0]

    if prediction == -1:
        status = "Anomaly"
    else:
        status = "Normal"

    return {
        "amount": amount,
        "category": category,
        "status": status,
        "anomaly_score": score,
        "category_average": stats["category_mean"],
        "category_zscore": category_zscore
    }
```

```python
def explain_expense(amount, category):
    result = analyze_expense(amount, category)

    if result["status"] == "Anomaly":
        print(
            f"⚠️ Potentially unusual expense: ₹{amount} "
            f"spent on {category}."
        )
        print(
            f"Your average {category} expense is "
            f"₹{result['category_average']:.0f}."
        )
        print(
            f"This transaction is "
            f"{result['category_zscore']:.1f} standard deviations "
            f"above your category average."
        )
    else:
        print(
            f"✅ This ₹{amount} {category} expense "
            f"looks normal for your spending pattern."
        )
```

```python
import pandas as pd
import numpy as np

np.random.seed(42)

n = 1000

categories = [
    "Food",
    "Transport",
    "Shopping",
    "Entertainment",
    "Education",
    "Healthcare",
    "Bills"
]

category_names = {
    "Food": ["Canteen", "Swiggy", "Zomato", "Restaurant"],
    "Transport": ["Bus", "Uber", "Ola", "Metro"],
    "Shopping": ["Amazon", "Myntra", "Local Store"],
    "Entertainment": ["Netflix", "PVR", "Gaming"],
    "Education": ["Udemy", "Books", "College"],
    "Healthcare": ["Pharmacy", "Clinic"],
    "Bills": ["Electricity", "Internet", "Mobile"]
}

category_amounts = {
    "Food": (80, 500),
    "Transport": (30, 300),
    "Shopping": (200, 2000),
    "Entertainment": (100, 1000),
    "Education": (100, 1500),
    "Healthcare": (100, 1200),
    "Bills": (200, 1500)
}

dates = pd.date_range(
    start="2026-07-01",
    end="2026-09-30",
    periods=n
)

rows = []

for date in dates:
    category = np.random.choice(categories)
    name = np.random.choice(category_names[category])

    low, high = category_amounts[category]
    amount = np.random.randint(low, high + 1)

    account = np.random.choice(
        ["Savings", "Cash", "Credit"],
        p=[0.6, 0.25, 0.15]
    )

    rows.append([
        date,
        name,
        amount,
        category,
        account,
        "Expense",
        ""
    ])

df = pd.DataFrame(
    rows,
    columns=[
        "date",
        "name",
        "amount",
        "category",
        "account",
        "type",
        "note"
    ]
)

print(df.shape)
```

```python
from sklearn.ensemble import IsolationForest

base_df = df.copy()
normal_df = base_df.copy()

category_stats = normal_df.groupby("category")["amount"].agg(
    category_mean="mean",
    category_std="std"
)

train_df = normal_df.merge(
    category_stats,
    on="category",
    how="left"
)

train_df["category_zscore"] = (
    train_df["amount"] - train_df["category_mean"]
) / train_df["category_std"]

X_train = train_df[
    ["amount", "category_zscore"]
]

model_v2 = IsolationForest(
    contamination=0.05,
    random_state=42
)

model_v2.fit(X_train)

print("Model trained successfully!")
```

```python
def analyze_expense(amount, category):

    stats = category_stats.loc[category]

    category_zscore = (
        amount - stats["category_mean"]
    ) / stats["category_std"]

    features = pd.DataFrame(
        [[amount, category_zscore]],
        columns=["amount", "category_zscore"]
    )

    prediction = model_v2.predict(features)[0]

    return "Anomaly" if prediction == -1 else "Normal"


print("₹200 Food:", analyze_expense(200, "Food"))
print("₹1500 Food:", analyze_expense(1500, "Food"))
```
